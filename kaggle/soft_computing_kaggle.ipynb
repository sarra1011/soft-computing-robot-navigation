```python
# --- write every source file of the project ---
import os, zlib, base64, ast
ROOT = '/kaggle/working/soft-computing-nav' if os.path.isdir('/kaggle/working') else os.path.abspath('soft-computing-nav')
BLOB = ''
files = ast.literal_eval(zlib.decompress(base64.b64decode(BLOB)).decode('utf-8'))
for rel, text in files.items():
    path = os.path.join(ROOT, rel)
    os.makedirs(os.path.dirname(path), exist_ok=True)
    with open(path, 'w', encoding='utf-8') as f: f.write(text)
os.chdir(ROOT)
print(len(files), 'files written to', ROOT)
```
